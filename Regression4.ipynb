```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

df = pd.read_csv('/content/home-rental.csv')
print("First 5 rows of dataset:")
print(df.head())

x=df[["size"]]
y=df[["rent_amount"]]

x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.2,random_state=42
)

model=LinearRegression()
model.fit(x_train,y_train)

y_pred=model.predict(x_test)

mse=mean_squared_error(y_test,y_pred)
print("Mean Squard Eror:",mse)

plt.scatter(x,y,color="red")
plt.plot(x,model.predict(x),color="blue")
plt.xlabel("size")
plt.ylabel("rent_amount")
plt.title("home rent analysis")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering

df = pd.read_csv('/content/archive (1).zip')

x = df[['Annual Income (k$)','Spending Score (1-100)']]

agg_model = AgglomerativeClustering(n_clusters=3)

df['cluster'] = agg_model.fit_predict(x)

plt.scatter(x['Annual Income (k$)'],x['Spending Score (1-100)'],c=df['cluster'],cmap='rainbow')
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Agglomerative Clustering Example")
plt.show()

df.to_csv("agglo_clustered data.csv",index=False)
print("Agglomerative clustered dataset saved as 'agglo_clusterd_data.csv'")
```
